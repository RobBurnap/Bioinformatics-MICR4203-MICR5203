## 2. Locate or create the standard course workspace

The setup recognizes both supported course-folder locations automatically. Students normally use `MyDrive/BIOINFO4-5203-F26/`; the instructor copy may use `MyDrive/Teaching/BIOINFO4-5203-F26/`.

This cell creates the standard top-level folders (`Notebooks`, `Data`, `Outputs`, and `Portfolio`) if they are missing, together with the NB00-specific `Data` and `Outputs` folders.


```python
from pathlib import Path

COURSE_FOLDER_NAME = "BIOINFO4-5203-F26"
NOTEBOOK_ID = "NB00_workspace_setup"

candidate_course_dirs = [
    Path("/content/drive/MyDrive") / COURSE_FOLDER_NAME,
    Path("/content/drive/MyDrive/Teaching") / COURSE_FOLDER_NAME,
]
existing_course_dirs = [path for path in candidate_course_dirs if path.exists()]

if existing_course_dirs:
    COURSE_DIR = existing_course_dirs[0]
else:
    COURSE_DIR = candidate_course_dirs[0]
    COURSE_DIR.mkdir(parents=True, exist_ok=True)
    print("Created the standard student course folder.")

NOTEBOOKS_DIR = COURSE_DIR / "Notebooks"
DATA_ROOT = COURSE_DIR / "Data"
OUTPUT_ROOT = COURSE_DIR / "Outputs"
PORTFOLIO_DIR = COURSE_DIR / "Portfolio"
PORTFOLIO_SELECTED_DIR = PORTFOLIO_DIR / "selected_final_results"

for folder in [NOTEBOOKS_DIR, DATA_ROOT, OUTPUT_ROOT, PORTFOLIO_DIR, PORTFOLIO_SELECTED_DIR]:
    folder.mkdir(parents=True, exist_ok=True)

DATA_DIR = DATA_ROOT / NOTEBOOK_ID
OUTPUT_DIR = OUTPUT_ROOT / NOTEBOOK_ID
DATA_DIR.mkdir(parents=True, exist_ok=True)
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

print("Course folder:    ", COURSE_DIR)
print("Notebooks folder: ", NOTEBOOKS_DIR)
print("Input folder:     ", DATA_DIR)
print("Output folder:    ", OUTPUT_DIR)
print("Portfolio folder: ", PORTFOLIO_DIR)
```

## 3. Preflight check

Confirm that the complete course workspace exists before continuing. NB00 does not require an external input file, but it also reports any files already present in its `Data` folder.


```python
required_folders = {
    "Course root": COURSE_DIR,
    "Notebooks": NOTEBOOKS_DIR,
    "Data": DATA_ROOT,
    "Outputs": OUTPUT_ROOT,
    "Portfolio": PORTFOLIO_DIR,
    "Portfolio/selected_final_results": PORTFOLIO_SELECTED_DIR,
    f"Data/{NOTEBOOK_ID}": DATA_DIR,
    f"Outputs/{NOTEBOOK_ID}": OUTPUT_DIR,
}

print("Workspace folder check:")
for label, path in required_folders.items():
    status = "OK" if path.is_dir() else "MISSING"
    print(f" - {label}: {status}")

missing_folders = [label for label, path in required_folders.items() if not path.is_dir()]
if missing_folders:
    raise FileNotFoundError("Missing required folder(s): " + ", ".join(missing_folders))

input_files = sorted(p.name for p in DATA_DIR.iterdir() if p.is_file())
print(f"\nFound {len(input_files)} file(s) currently in {DATA_DIR.name}:")
for name in input_files:
    print(" -", name)
if not input_files:
    print(" - none (this is expected for a new NB00 workspace)")
```

## 4. Demonstrate temporary versus persistent storage

Colab's local `/content/` area belongs to the current runtime and can disappear when the session ends. Files written under `/content/drive/MyDrive/` are stored in Google Drive and persist after the Colab runtime disconnects.

The next cell creates one small temporary demonstration file and the two required persistent output files.


```python
import platform
import sys
from datetime import datetime

# A temporary file inside the current Colab runtime.
TEMP_FILE = Path("/content/NB00_temporary_storage_demo.txt")
TEMP_FILE.write_text(
    "This file is stored only in the current Colab runtime and may disappear when the runtime ends.\n"
)

# Persistent reproducibility summary saved in Google Drive.
environment_summary = [
    "NB00 environment summary",
    f"Created: {datetime.now().astimezone().isoformat(timespec='seconds')}",
    f"Python version: {sys.version.split()[0]}",
    f"Platform: {platform.platform()}",
    f"Course folder: {COURSE_DIR}",
    f"Notebooks folder: {NOTEBOOKS_DIR}",
    f"Data folder: {DATA_DIR}",
    f"Output folder: {OUTPUT_DIR}",
]

ENVIRONMENT_FILE = OUTPUT_DIR / "environment_summary.txt"
ENVIRONMENT_FILE.write_text("\n".join(environment_summary) + "\n")

# Persistent folder verification saved in Google Drive.
folder_check_lines = ["NB00 workspace folder check"]
for label, path in required_folders.items():
    status = "OK" if path.is_dir() else "MISSING"
    folder_check_lines.append(f"{label}: {status} | {path}")

FOLDER_CHECK_FILE = OUTPUT_DIR / "folder_check.txt"
FOLDER_CHECK_FILE.write_text("\n".join(folder_check_lines) + "\n")

print("Temporary Colab file:")
print(" ", TEMP_FILE)
print("\nPersistent Google Drive files:")
print(" ", ENVIRONMENT_FILE)
print(" ", FOLDER_CHECK_FILE)
print("\nThe file under /content/ is temporary; the files under /content/drive/MyDrive/ are persistent.")
```
